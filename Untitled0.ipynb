```python
from collections import Counter, defaultdict

split_counts = Counter()
label_counts = Counter()
modality_counts = Counter()
examples = defaultdict(list)

for item in metadata:
    split = item.get("split", "unknown")
    split_counts[split] += 1

    # Provisional binary label: 1 = fake, 0 = real
    is_fake = (
        item.get("n_fakes", 0) > 0
        or item.get("modify_video", False)
        or item.get("modify_audio", False)
    )

    label_counts[(split, int(is_fake))] += 1

    if item.get("modify_video", False):
        modality_counts[(split, "video_modified")] += 1

    if item.get("modify_audio", False):
        modality_counts[(split, "audio_modified")] += 1

    if len(examples[split]) < 3:
        examples[split].append({
            "file": item.get("file"),
            "n_fakes": item.get("n_fakes"),
            "modify_video": item.get("modify_video"),
            "modify_audio": item.get("modify_audio"),
            "original": item.get("original")
        })

print("Records by split:", dict(split_counts))
print("\nLabels by split (0=real, 1=fake):")
for key, count in sorted(label_counts.items()):
    print(key, ":", count)

print("\nModality modification counts:")
for key, count in sorted(modality_counts.items()):
    print(key, ":", count)

print("\nExample records:")
for split, rows in examples.items():
    print(split, rows)
```

```python
import os
import cv2
import numpy as np
import librosa
import torch
import subprocess

sample_path = os.path.join(DATASET_DIR, "test/000001.mp4")

# 1. Extract audio as a mono waveform at 16 kHz
audio_path = "/tmp/lavdf_sample.wav"

subprocess.run(
    [
        "ffmpeg", "-y", "-i", sample_path,
        "-vn", "-ac", "1", "-ar", "16000",
        "-c:a", "pcm_s16le", audio_path
    ],
    check=True,
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL
)

audio, sr = librosa.load(audio_path, sr=16000, mono=True)

# 2. Extract 8 evenly spaced video frames
cap = cv2.VideoCapture(sample_path)
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

indices = np.linspace(
    0, total_frames - 1, num=min(8, total_frames), dtype=int
)

frames = []

for idx in indices:
    cap.set(cv2.CAP_PROP_POS_FRAMES, int(idx))
    success, frame = cap.read()

    if success:
        frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        frame = cv2.resize(frame, (224, 224))
        frames.append(frame)

cap.release()

# 3. Convert to tensors
audio_tensor = torch.from_numpy(audio.copy()).float()
video_tensor = torch.from_numpy(np.stack(frames)).permute(0, 3, 1, 2).float() / 255.0

print("Sample rate:", sr)
print("Audio duration:", round(len(audio) / sr, 3), "seconds")
print("Audio tensor shape:", tuple(audio_tensor.shape))
print("Video tensor shape:", tuple(video_tensor.shape))
print("Audio finite:", torch.isfinite(audio_tensor).all().item())
print("Video finite:", torch.isfinite(video_tensor).all().item())
print("Extracted frames:", len(frames))
```

```python

import os
import cv2
import json
import torch
import librosa
import numpy as np
import subprocess

from torch.utils.data import Dataset, DataLoader

# Dataset configuration
DATASET_DIR = (
    "/root/.cache/kagglehub/datasets/"
    "elin75/localized-audio-visual-deepfake-dataset-lav-df/"
    "versions/1/LAV-DF"
)

with open(
    os.path.join(DATASET_DIR, "metadata.min.json"),
    "r",
    encoding="utf-8"
) as f:
    metadata = json.load(f)


class LAVDFDataset(Dataset):
    def __init__(self, records, num_frames=8):
        self.records = records
        self.num_frames = num_frames

    def __len__(self):
        return len(self.records)

    def __getitem__(self, idx):
        item = self.records[idx]
        path = os.path.join(DATASET_DIR, item["file"])

        # Binary and modality labels
        audio_fake = int(bool(item["modify_audio"]))
        video_fake = int(bool(item["modify_video"]))
        label = int(audio_fake or video_fake)

        # Decode audio at 16 kHz
        result = subprocess.run(
            [
                "ffmpeg", "-v", "error", "-i", path,
                "-vn", "-ac", "1", "-ar", "16000",
                "-f", "f32le", "-"
            ],
            capture_output=True,
            check=True
        )

        audio = np.frombuffer(
            result.stdout, dtype=np.float32
        ).copy()

        # Fixed-length audio: 4 seconds
        target_length = 16000 * 4

        if len(audio) < target_length:
            audio = np.pad(
                audio, (0, target_length - len(audio))
            )
        else:
            audio = audio[:target_length]

        # Log-mel spectrogram
        mel = librosa.feature.melspectrogram(
            y=audio,
            sr=16000,
            n_fft=400,
            hop_length=160,
            n_mels=80,
            power=2.0
        )

        log_mel = librosa.power_to_db(
            mel, ref=np.max
        )

        # Normalize each sample
        log_mel = (
            (log_mel - log_mel.mean())
            / (log_mel.std() + 1e-6)
        )

        # Decode evenly spaced video frames
        cap = cv2.VideoCapture(path)
        total_frames = int(
            cap.get(cv2.CAP_PROP_FRAME_COUNT)
        )

        if total_frames < 1:
            cap.release()
            raise RuntimeError(f"No video frames: {path}")

        indices = np.linspace(
            0, total_frames - 1,
            self.num_frames, dtype=int
        )

        frames = []

        for frame_idx in indices:
            cap.set(cv2.CAP_PROP_POS_FRAMES, int(frame_idx))
            ok, frame = cap.read()

            if not ok:
                cap.release()
                raise RuntimeError(
                    f"Frame decoding failed: {path}"
                )

            frame = cv2.cvtColor(
                frame, cv2.COLOR_BGR2RGB
            )
            frame = cv2.resize(frame, (224, 224))
            frames.append(frame)

        cap.release()

        video = (
            torch.from_numpy(np.stack(frames))
            .permute(0, 3, 1, 2)
            .float() / 255.0
        )

        return {
            "audio": torch.from_numpy(log_mel).float().unsqueeze(0),
            "video": video,
            "label": torch.tensor(label, dtype=torch.long),
            "audio_label": torch.tensor(audio_fake, dtype=torch.long),
            "video_label": torch.tensor(video_fake, dtype=torch.long),
        }


# Use only 8 records to validate the pipeline
train_records = [
    item for item in metadata
    if item["split"] == "train"
][:8]

pilot_dataset = LAVDFDataset(train_records)
pilot_loader = DataLoader(
    pilot_dataset, batch_size=2, shuffle=False,
    num_workers=0
)

batch = next(iter(pilot_loader))

for key, value in batch.items():
    print(key, tuple(value.shape), value.dtype)

print("All audio finite:", torch.isfinite(batch["audio"]).all().item())
print("All video finite:", torch.isfinite(batch["video"]).all().item())
print("Binary labels:", batch["label"].tolist())
print("Audio labels:", batch["audio_label"].tolist())
print("Video labels:", batch["video_label"].tolist())
```

```python

from collections import defaultdict
from torch.utils.data import DataLoader

# Group training records into four manipulation categories
groups = defaultdict(list)

for item in metadata:
    if item["split"] != "train":
        continue

    audio_fake = bool(item["modify_audio"])
    video_fake = bool(item["modify_video"])

    if not audio_fake and not video_fake:
        category = "real"
    elif audio_fake and not video_fake:
        category = "audio_only"
    elif video_fake and not audio_fake:
        category = "video_only"
    else:
        category = "both"

    groups[category].append(item)

# Select one record from each category
selected = [
    groups[category][0]
    for category in ["real", "audio_only", "video_only", "both"]
]

print("Selected examples:")
for item in selected:
    print(
        item["file"],
        "| audio modified:", item["modify_audio"],
        "| video modified:", item["modify_video"],
        "| n_fakes:", item["n_fakes"]
    )

# Decode all four examples
category_dataset = LAVDFDataset(selected)
category_loader = DataLoader(
    category_dataset, batch_size=4, shuffle=False, num_workers=0
)

batch = next(iter(category_loader))

print("\nBinary labels:", batch["label"].tolist())
print("Audio labels:", batch["audio_label"].tolist())
print("Video labels:", batch["video_label"].tolist())
print("Audio shape:", tuple(batch["audio"].shape))
print("Video shape:", tuple(batch["video"].shape))
```

```python

import time
from torch.utils.data import DataLoader

benchmark_dataset = LAVDFDataset(
    [item for item in metadata if item["split"] == "train"][:32]
)

benchmark_loader = DataLoader(
    benchmark_dataset,
    batch_size=4,
    shuffle=False,
    num_workers=0
)

start = time.perf_counter()
batch_count = 0

for batch in benchmark_loader:
    batch_count += 1

elapsed = time.perf_counter() - start

print("Samples processed:", batch_count * 4)
print("Batches processed:", batch_count)
print("Elapsed seconds:", round(elapsed, 2))
print("Seconds per batch:", round(elapsed / batch_count, 2))
print("Samples per second:", round((batch_count * 4) / elapsed, 2))
```

```python

import time
from torch.utils.data import DataLoader

records = [
    item for item in metadata
    if item["split"] == "train"
][:32]

for workers in [0, 2, 4]:
    loader = DataLoader(
        LAVDFDataset(records),
        batch_size=4,
        shuffle=False,
        num_workers=workers,
        persistent_workers=(workers > 0)
    )

    start = time.perf_counter()
    samples = 0

    for batch in loader:
        samples += len(batch["label"])

    elapsed = time.perf_counter() - start

    print(
        f"Workers: {workers} | "
        f"Time: {elapsed:.2f}s | "
        f"Throughput: {samples / elapsed:.2f} samples/sec"
    )
```
